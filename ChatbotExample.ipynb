```python
from datetime import datetime
from zoneinfo import ZoneInfo

while True:
    user = input("Hi! How can i help you? / Hai! Ada apa yang boleh saya bantu? (type 'exit'/'keluar' to quit): ")

    # Update the time dynamically each turn of the loop
    kl_time = datetime.now(ZoneInfo("Asia/Kuala_Lumpur"))
    user_lower = user.lower()

    if any(word in user_lower for word in ["exit", "quit", "keluar"]):
        print("Goodbye! / Jumpa lagi!")
        break
    elif "weather" in user_lower or "cuaca" in user_lower:
        print("The weather today is sunny! / Cuaca hari ini cerah!")
    elif any(word in user_lower for word in ["time", "masa", "pukul"]):
        formatted_time = kl_time.strftime("%I:%M %p")
        print(f"The current time in Kuala Lumpur is {formatted_time}. / Waktu sekarang di Kuala Lumpur ialah {formatted_time}.")
    else:
        print("Sorry, I don't understand. / Maaf, saya tidak faham.")
```
